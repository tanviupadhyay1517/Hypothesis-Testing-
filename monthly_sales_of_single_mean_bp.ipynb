# Normality Check

```python
# Shapiro test

shapiro(df['Monthly_Sales'])
```

```python
# qq plot
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot

qqplot(df['Monthly_Sales'],line='s')
plt.show()
```

# t test for single sample

```python
ttest1 = ttest_1samp(df['Monthly_Sales'],500,alternative='greater')
ttest1
```
