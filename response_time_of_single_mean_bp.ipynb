# Normality Check

```python
# Shapiro test

shapiro(df['Response_Time'])
```

```python
# qq plot
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot

qqplot(df['Response_Time'],line='s')
plt.show()
```

# t test for single sample

```python
ttest1 = ttest_1samp(df['Response_Time'],24,alternative='less')
ttest1
```
