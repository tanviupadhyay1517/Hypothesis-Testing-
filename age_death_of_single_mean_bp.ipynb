# Normality Check

```python
# Shapiro test

shapiro(df['Age'])
```

```python
# qq plot
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot

qqplot(df['Age'],line='s')
plt.show()
```

# t test for single sample

```python
ttest1 = ttest_1samp(df['Age'],25,alternative='less')
ttest1
```
