**Case Study:**
17 families of mentally retarded children participated in the training program. The test assesses knowledge of behavior modification in the primary parent of each of 17 families. The data of the pre and post training scores made by the
Primary parents on test are recorded in retarded_children.xlsx file. May we conclude that the training program increased knowledge of behavior modification principles?


```python
# Import necessary packages
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
# Read data
df1 = pd.read_excel('/content/drive/MyDrive/Classroom/FTN Batch 10/T test Two sample/retarded_children.xlsx')
```

# Exploratory Analysis

```python
df1.columns
```

```python
df1.head()                                  # Viewing first few observations of the data
```

```python
df1.describe()                              # Descriptive statistics
```

```python
# Taking difference
df1['errors_diff'] = df1['Post_scores'] - df1['Pre_scores']
df1['errors_diff'].describe()
```

```python
# Boxplot
plt.figure(figsize=(8, 6))
plt.boxplot(df1['errors_diff'])
plt.show()
```

```python
# Checking Normality using Shapiro test
Shapiro1=stats.shapiro(df1['errors_diff'])
Shapiro1
```

```python
# Checking Normality using Q-Q plot
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot

qqplot(df1['errors_diff'],line='s')
plt.show()
```

```python
# Paired t test / t test for dependent samples

ttest=stats.ttest_rel(df1['Post_scores'] , df1['Pre_scores'], alternative ='greater')
ttest
```

```python
print("Average errors before training:", df1['Pre_scores'].mean())
print("Average errors after training :", df1['Post_scores'].mean())
```
