Case Study

# Normality Check

```python
# Shapiro test

shapiro(df['Head_Circumference '])
```

```python
# qq plot
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot

qqplot(df['Head_Circumference '],line='s')
plt.show()
```

# t test for single sample

```python
ttest1 = ttest_1samp(df['Head_Circumference '],34.5,alternative='two-sided')
ttest1
```
