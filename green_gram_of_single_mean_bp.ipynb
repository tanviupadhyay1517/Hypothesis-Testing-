Case Study

The yield (quintals/hectare) were recorded in the file Green_gram.xlsx. Does the result confirm to the expectation?

# Normality Check

```python
# Shapiro test

shapiro(df['Yield '])
```

```python
# qq plot
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot

qqplot(df['Yield '],line='s')
plt.show()
```

# t test for single sample

```python
ttest1 = ttest_1samp(df['Yield '],12.0,alternative='two-sided')
ttest1
```
