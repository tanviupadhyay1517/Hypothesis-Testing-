# Normality Check

```python
# Shapiro test

shapiro(df['Processing_Time'])
```

```python
# qq plot
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot

qqplot(df['Processing_Time'],line='s')
plt.show()
```

# t test for single sample

```python
ttest1 = ttest_1samp(df['Processing_Time'],40,alternative='greater')
ttest1
```
