# Normality test using Shapiro test

````python
Shapiro1 = stats.shapiro(df.Yield[df.Fertilizer == 1])

print("Shapiro-Wilk test for Fertilizer 1:", Shapiro1)
````

````python
Shapiro2 = stats.shapiro(df.Yield[df.Fertilizer == 2])

print("Shapiro-Wilk test for Fertilizer 1:", Shapiro2)
````

# Normality check using Q-Q plot

````python
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot
````

````python
qqplot(df[df['Fertilizer'] == 1]['Yield'], line='s')
plt.show()
````

````python
qqplot(df[df['Fertilizer'] == 2]['Yield'], line='s')
plt.show()
````

## Testing Equality of Variances

````python
# Levene test using Center as Mean
stats.levene(df.Yield[df.Fertilizer == 1],df.Yield[df.Fertilizer == 2],center='mean')
````
