The birth rate (per 1000 residential population) for random samples of counties in California and Maine. Can we conclude that the average birth rate of both counties equal? (The data is available in birthrates.xlsx)

# Normality test using Shapiro test

```python
Shapiro1 = stats.shapiro(df.birthrate[df.county == 'California'])
print("Shapiro-Wilk test for county California:", Shapiro1)
```

```python
Shapiro2 = stats.shapiro(df.birthrate[df.county == 'Maine'])
print("Shapiro-Wilk test for country Maine:", Shapiro2)
```

# Normality check using Q-Q plot

```python
from statsmodels.graphics.gofplots import qqplot
from matplotlib import pyplot
```

```python
qqplot(df[df['county'] == 'California']['birthrate'], line='s')
plt.show()
```

```python
qqplot(df[df['county'] == 'Maine']['birthrate'], line='s')
plt.show()
```

## Testing Equality of Variances

```python
# Levene test using Center as Mean
stats.levene(df.birthrate[df.county == 'California'],df.birthrate[df.county == 'Maine'],center='mean')
```
